# Data Access via OPeNDAP

```python
import numpy as np
import pandas as pd
import xarray as xr 
```

```python
# location of interest 
# exact location of the beach 
lon_sl = 81.77
lat_sl = 7.64
```

## Accessing single file

```python
url = 'https://data-cbr.csiro.au/thredds/dodsC/catch_all/oa-cmip6-wave/UniMelb-CSIRO_CMIP6_projections/historical/ACCESS-CM2/CDFAC108/ww3_ounf_glout/ww3.201412_hs.nc'
ds = xr.open_dataset(url, engine='pydap')

ts = ds['hs'].sel(longitude=lon_sl, latitude=lat_sl, method='nearest')
print(ts)
```
